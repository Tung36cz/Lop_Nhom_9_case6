```python
import numpy as np
import csv

path2 = r"station_bac.csv"

# đọc header
header = np.genfromtxt(path2, delimiter=",", max_rows=1, dtype=str, encoding='utf-8')

# tìm index cột (hỗ trợ 'rain' hoặc 'rainfall')
try:
    idx_date = list(header).index('date')
except ValueError:
    idx_date = 0  # mặc định date ở cột 0 nếu không thấy 'date' trong header

try:
    idx_temp = list(header).index('temp')
except ValueError:
    idx_temp = 1

try:
    idx_hum = list(header).index('humidity')
except ValueError:
    idx_hum = 2

# rain có thể tên 'rain' hoặc 'rainfall'
if 'rain' in list(header):
    idx_rain = list(header).index('rain')
elif 'rainfall' in list(header):
    idx_rain = list(header).index('rainfall')
else:
    idx_rain = 3

cols = np.array([idx_temp, idx_hum, idx_rain])

# đọc toàn bộ dữ liệu dưới dạng chuỗi để có thể lấy date (chuỗi) và các cột số
data_all = np.genfromtxt(path2, delimiter=",", dtype=str, encoding='utf-8', skip_header=1)
if data_all.ndim == 1:
    data_all = data_all.reshape(1, -1)

# tách date và cột số
date_col = data_all[:, idx_date]
selected = data_all[:, cols].astype(float)  # các cột số

# tính biến động bằng np.diff
diffs = np.abs(np.diff(selected, axis=0))

th_temp, th_hum, th_rain = 8.0, 15.0, 40.0
flags_temp = diffs[:, 0] > th_temp
flags_hum = diffs[:, 1] > th_hum
flags_rain = diffs[:, 2] > th_rain
any_flag = flags_temp | flags_hum | flags_rain

n = data_all.shape[0]
tinh_trang = np.full(n, 'Bình thường', dtype=object)
tinh_trang[0] = 'Không xác định'
tinh_trang[1:][any_flag] = 'Biến động mạnh'

# không tạo chi_tiet nữa

# in ra (có date) — không in chi_tiet
for i in range(n):
    print(i, date_col[i], selected[i], tinh_trang[i])

# ghi CSV: thêm cột date trước các cột số, không ghi chi_tiet

try:
    out_header = ['station', header[idx_date]] + list(header[cols]) + ['tinh_trang']
except Exception:
    out_header = ['station', 'date', 'temp', 'humidity', 'rain', 'tinh_trang']

with open(r"ket_qua_case_6.csv", "w", newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f,fieldnames=["Trạm nghiên cứu miền Bắc"])
    writer.writeheader()#Lệnh để ghi lên hàng đầu của bảng
    writer = csv.writer(f)
    writer.writerow(out_header)
    for i in range(n):
        row_values = [date_col[i]] + [str(x) for x in selected[i]] + [tinh_trang[i]]
        writer.writerow(row_values)


path2 = r"station_nam.csv"

# đọc header
header = np.genfromtxt(path2, delimiter=",", max_rows=1, dtype=str, encoding='utf-8')

# tìm index cột (hỗ trợ 'rain' hoặc 'rainfall')
try:
    idx_date = list(header).index('date')
except ValueError:
    idx_date = 0  # mặc định date ở cột 0 nếu không thấy 'date' trong header

try:
    idx_temp = list(header).index('temp')
except ValueError:
    idx_temp = 1

try:
    idx_hum = list(header).index('humidity')
except ValueError:
    idx_hum = 2

# rain có thể tên 'rain' hoặc 'rainfall'
if 'rain' in list(header):
    idx_rain = list(header).index('rain')
elif 'rainfall' in list(header):
    idx_rain = list(header).index('rainfall')
else:
    idx_rain = 3

cols = np.array([idx_temp, idx_hum, idx_rain])

# đọc toàn bộ dữ liệu dưới dạng chuỗi để có thể lấy date (chuỗi) và các cột số
data_all = np.genfromtxt(path2, delimiter=",", dtype=str, encoding='utf-8', skip_header=1)
if data_all.ndim == 1:
    data_all = data_all.reshape(1, -1)

# tách date và cột số
date_col = data_all[:, idx_date]
selected = data_all[:, cols].astype(float)  # các cột số

# tính biến động bằng np.diff
diffs = np.abs(np.diff(selected, axis=0))

th_temp, th_hum, th_rain = 8.0, 15.0, 40.0
flags_temp = diffs[:, 0] > th_temp
flags_hum = diffs[:, 1] > th_hum
flags_rain = diffs[:, 2] > th_rain
any_flag = flags_temp | flags_hum | flags_rain

n = data_all.shape[0]
tinh_trang = np.full(n, 'Bình thường', dtype=object)
tinh_trang[0] = 'Không xác định'
tinh_trang[1:][any_flag] = 'Biến động mạnh'

# không tạo chi_tiet nữa

# in ra (có date) — không in chi_tiet
for i in range(n):
    print(i, date_col[i], selected[i], tinh_trang[i])

# ghi CSV: thêm cột date trước các cột số, không ghi chi_tiet

# ghi tiếp (append), không ghi lại header
with open(r"ket_qua_case_6.csv", "a", newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f,fieldnames=["Trạm nghiên cứu miền Nam"])
    writer.writeheader()#Lệnh để ghi lên hàng đầu của bảng
    writer = csv.writer(f)
    for i in range(n):
        row_values = [ date_col[i]] + [str(x) for x in selected[i]] + [tinh_trang[i]]
        writer.writerow(row_values)
```
