```python
import numpy as np
data_bac_raw = np.genfromtxt('station_bac.csv', delimiter=',', dtype=str, skip_header=1)
data_nam_raw = np.genfromtxt('station_nam.csv', delimiter=',', dtype=str, skip_header=1)
# Trạm Bắc
date_bac = np.array(data_bac_raw[:, 0], dtype='datetime64[D]')
temp_bac = data_bac_raw[:, 1].astype(float)
hum_bac  = data_bac_raw[:, 2].astype(float)
rain_bac = data_bac_raw[:, 3].astype(float)

# Trạm Nam
date_nam = np.array(data_nam_raw[:, 0], dtype='datetime64[D]')
temp_nam = data_nam_raw[:, 1].astype(float)
hum_nam  = data_nam_raw[:, 2].astype(float)
rain_nam = data_nam_raw[:, 3].astype(float)
```

```python
#Mục tiêu 2
print("=== 5. SLICING 30 NGÀY ĐẦU, GIỮA, CUỐI ===\n")
n = len(data_bac_raw)
print(f"Tổng số ngày: {n}\n")

# 30 ngày đầu
first_30 = data_bac_raw[:30]
print(f"30 ngày đầu (trạm Bắc):\n{first_30[:5]}...\n")

# 30 ngày giữa
mid_start = (n - 30) // 2
mid_30 = data_bac_raw[mid_start:mid_start + 30]
print(f"30 ngày giữa (từ ngày {mid_start} đến {mid_start + 30}):\n{mid_30[:5]}...\n")

# 30 ngày cuối
last_30 = data_bac_raw[-30:]
print(f"30 ngày cuối:\n{last_30[:5]}...\n")

# === 6. BOOLEAN MASK PHÁT HIỆN BẤT THƯỜNG ===
print("=== 6. BOOLEAN MASK PHÁT HIỆN BẤT THƯỜNG ===\n")

# Trạm Bắc: nhiệt độ > 35°C
extreme_temp_bac = data_bac_raw[temp_bac > 35]
print(f"Ngày nhiệt độ > 35°C (Bắc): {len(extreme_temp_bac)} ngày")
if len(extreme_temp_bac) > 0:
    print(f"  Mẫu: {extreme_temp_bac[:3]}\n")
else:
    print("  Không có ngày nào\n")

# Trạm Nam: nhiệt độ > 35°C
extreme_temp_nam = data_nam_raw[temp_nam > 35]
print(f"Ngày nhiệt độ > 35°C (Nam): {len(extreme_temp_nam)} ngày")
if len(extreme_temp_nam) > 0:
    print(f"  Mẫu: {extreme_temp_nam[:3]}\n")
else:
    print("  Không có ngày nào\n")

# Trạm Bắc: độ ẩm < 30%
low_hum_bac = data_bac_raw[hum_bac < 30]
print(f"Ngày độ ẩm < 30% (Bắc): {len(low_hum_bac)} ngày")
if len(low_hum_bac) > 0:
    print(f"  Mẫu: {low_hum_bac[:3]}\n")
else:
    print("  Không có ngày nào\n")

# Trạm Nam: độ ẩm < 30%
low_hum_nam = data_nam_raw[hum_nam < 30]
print(f"Ngày độ ẩm < 30% (Nam): {len(low_hum_nam)} ngày")
if len(low_hum_nam) > 0:
    print(f"  Mẫu: {low_hum_nam[:3]}\n")
else:
    print("  Không có ngày nào\n")

# Trạm Bắc: mưa > 50 mm
heavy_rain_bac = data_bac_raw[rain_bac > 50]
print(f"Ngày mưa > 50mm (Bắc): {len(heavy_rain_bac)} ngày")
if len(heavy_rain_bac) > 0:
    print(f"  Mẫu: {heavy_rain_bac[:3]}\n")
else:
    print("  Không có ngày nào\n")

# Trạm Nam: mưa > 50 mm
heavy_rain_nam = data_nam_raw[rain_nam > 50]
print(f"Ngày mưa > 50mm (Nam): {len(heavy_rain_nam)} ngày")
if len(heavy_rain_nam) > 0:
    print(f"  Mẫu: {heavy_rain_nam[:3]}\n")
else:
    print("  Không có ngày nào\n")

# === 7. BROADCASTING SO SÁNH 2 TRẠM ===
print("=== 7. BROADCASTING SO SÁNH 2 TRẠM ===\n")

# Chênh lệch nhiệt độ (Bắc - Nam)
temp_diff = temp_bac - temp_nam
print(f"Chênh lệch nhiệt độ (Bắc - Nam):")
print(f"  Trung bình: {np.mean(temp_diff):+.2f}°C")
print(f"  Max: {np.max(temp_diff):+.2f}°C")
print(f"  Min: {np.min(temp_diff):+.2f}°C")
print(f"  5 ngày đầu: {temp_diff[:5]}\n")

# Chênh lệch độ ẩm (Bắc - Nam)
humid_diff = hum_bac - hum_nam
print(f"Chênh lệch độ ẩm (Bắc - Nam):")
print(f"  Trung bình: {np.mean(humid_diff):+.2f}%")
print(f"  Max: {np.max(humid_diff):+.2f}%")
print(f"  Min: {np.min(humid_diff):+.2f}%\n")

# Chênh lệch mưa (Bắc - Nam)
rain_diff = rain_bac - rain_nam
print(f"Chênh lệch mưa (Bắc - Nam):")
print(f"  Trung bình: {np.mean(rain_diff):+.2f}mm")
print(f"  Max: {np.max(rain_diff):+.2f}mm")
print(f"  Min: {np.min(rain_diff):+.2f}mm\n")

# === 8. FANCY INDEXING: TOP-10 NGÀY ===
print("=== 8. FANCY INDEXING: TOP-10 NGÀY ===\n")

# Top-10 ngày nhiệt độ cao nhất (Bắc)
top10_temp_bac_idx = np.argsort(temp_bac)[-10:][::-1]
top10_temp_bac = data_bac_raw[top10_temp_bac_idx]
print(f"Top-10 ngày nhiệt độ cao nhất (Bắc):")
for i, idx in enumerate(top10_temp_bac_idx, 1):
    print(f"  {i}. Ngày {idx+1}: {temp_bac[idx]:.2f}°C")
print()

# Top-10 ngày nhiệt độ cao nhất (Nam)
top10_temp_nam_idx = np.argsort(temp_nam)[-10:][::-1]
top10_temp_nam = data_nam_raw[top10_temp_nam_idx]
print(f"Top-10 ngày nhiệt độ cao nhất (Nam):")
for i, idx in enumerate(top10_temp_nam_idx, 1):
    print(f"  {i}. Ngày {idx+1}: {temp_nam[idx]:.2f}°C")
print()

# Top-10 ngày mưa lớn nhất (Bắc)
top10_rain_bac_idx = np.argsort(rain_bac)[-10:][::-1]
top10_rain_bac = data_bac_raw[top10_rain_bac_idx]
print(f"Top-10 ngày mưa lớn nhất (Bắc):")
for i, idx in enumerate(top10_rain_bac_idx, 1):
    print(f"  {i}. Ngày {idx+1}: {rain_bac[idx]:.2f}mm")
print()

# Top-10 ngày mưa lớn nhất (Nam)
top10_rain_nam_idx = np.argsort(rain_nam)[-10:][::-1]
top10_rain_nam = data_nam_raw[top10_rain_nam_idx]
print(f"Top-10 ngày mưa lớn nhất (Nam):")
for i, idx in enumerate(top10_rain_nam_idx, 1):
    print(f"  {i}. Ngày {idx+1}: {rain_nam[idx]:.2f}mm")
```

```python
#Mục tiêu 4
def phan_tich_bien_dong(data_array, date_array, threshold, column_name, station_name):
    """
    Hàm tính toán biến động ngày-qua-ngày sử dụng np.diff
    """
    changes = np.diff(data_array)

    abs_changes = np.abs(changes)
    
    mask_fluctuation = abs_changes > threshold
    
    dates_fluctuation = date_array[1:][mask_fluctuation]
    values_fluctuation = changes[mask_fluctuation]
    
    print(f"\n[Trạm {station_name}] Biến động {column_name} (Ngưỡng > {threshold}):")
    print(f"- Tổng số ngày biến động mạnh: {len(dates_fluctuation)}")
    
    if len(dates_fluctuation) > 0:
        print("- Chi tiết 5 ngày đầu:")
        for date, val in zip(dates_fluctuation[:5], values_fluctuation[:5]):
            status = "Tăng" if val > 0 else "Giảm"
            print(f"  + {date}: {status} {abs(val):.2f}")

phan_tich_bien_dong(temp_bac, date_bac, 8, "Nhiệt độ", "Bắc")
phan_tich_bien_dong(hum_bac, date_bac, 15, "Độ ẩm", "Bắc")
phan_tich_bien_dong(rain_bac, date_bac, 40, "Lượng mưa", "Bắc")

print("-" * 30)
phan_tich_bien_dong(temp_nam, date_nam, 8, "Nhiệt độ", "Nam")
phan_tich_bien_dong(hum_nam, date_nam, 15, "Độ ẩm", "Nam")
phan_tich_bien_dong(rain_nam, date_nam, 40, "Lượng mưa", "Nam")
```

```python
#Mục tiêu 6
delta_temp_bac = np.diff(temp_bac)     
delta_humid_bac = np.diff(hum_bac)    
delta_rain_bac = np.diff(rain_bac)      

delta_temp_nam = np.diff(temp_nam)     
delta_humid_nam = np.diff(hum_nam)    
delta_rain_nam = np.diff(rain_nam)      

T_thres = 8.0
H_thres = 15.0
R_thres = 40.0

strong_temp_days_bac = np.where(np.abs(delta_temp_bac) > T_thres)[0]
strong_humid_days_bac = np.where(np.abs(delta_humid_bac) > H_thres)[0]
strong_rain_days_bac = np.where(np.abs(delta_rain_bac) > R_thres)[0]

strong_temp_days_nam = np.where(np.abs(delta_temp_nam) > T_thres)[0]
strong_humid_days_nam = np.where(np.abs(delta_humid_nam) > H_thres)[0]
strong_rain_days_nam = np.where(np.abs(delta_rain_nam) > R_thres)[0]

n_days_bac = len(delta_temp_bac)
tinh_trang_bac = np.full(n_days_bac, 'Bình thường', dtype=object)

n_days_nam = len(delta_temp_nam)
tinh_trang_nam = np.full(n_days_nam, 'Bình thường', dtype=object)

tinh_trang_bac[strong_temp_days_bac] = 'Biến động mạnh'
tinh_trang_bac[strong_humid_days_bac] = 'Biến động mạnh'
tinh_trang_bac[strong_rain_days_bac] = 'Biến động mạnh'

tinh_trang_nam[strong_temp_days_nam] = 'Biến động mạnh'
tinh_trang_nam[strong_humid_days_nam] = 'Biến động mạnh'
tinh_trang_nam[strong_rain_days_nam] = 'Biến động mạnh'
print("Trạm nghiên cứu khí tượng miền Bắc")
# IN chi tiết từng ngày
print("=== CHI TIẾT BIẾN ĐỘNG THEO NGÀY ===\n")
for i in range(n_days_bac):
    print(f"Ngày {i} → {i+1}:")
    print(f"  ΔT = {delta_temp_bac[i]:+.2f}°C | ΔH = {delta_humid_bac[i]:+.2f}% | ΔR = {delta_rain_bac[i]:+.2f}mm")
    print(f"  Tình trạng: {tinh_trang_bac[i]}")
    print()

# IN tóm tắt
print("=== BIẾN ĐỘNG NHIỆT ĐỘ (> 8°C) ===")
if len(strong_temp_days_bac) > 0:
    for i in strong_temp_days_bac:
        print(f"Ngày {i} → {i+1}: ΔT = {delta_temp_bac[i]:.2f}°C")
else:
    print("Không có ngày nào biến động nhiệt độ > 8°C")

print("\n=== BIẾN ĐỘNG ĐỘ ẨM (> 15%) ===")
if len(strong_humid_days_bac) > 0:
    for i in strong_humid_days_bac:
        print(f"Ngày {i} → {i+1}: ΔH = {delta_humid_bac[i]:.2f}%")
else:
    print("Không có ngày nào biến động độ ẩm > 15%")

print("\n=== BIẾN ĐỘNG LƯỢNG MƯA (> 40mm) ===")
if len(strong_rain_days_bac) > 0:
    for i in strong_rain_days_bac:
        print(f"Ngày {i} → {i+1}: ΔR = {delta_rain_bac[i]:.2f}mm")
else:
    print("Không có ngày nào biến động lượng mưa > 40mm")

# Tìm những ngày có bất kỳ biến động mạnh nào
all_strong_days_bac = np.unique(np.concatenate([strong_temp_days_bac, strong_humid_days_bac, strong_rain_days_bac]))
print(f"\n=== TỔNG HỢP ===")
print(f"Tổng ngày có biến động mạnh: {len(all_strong_days_bac)}")
print(f"Tổng ngày bình thường: {n_days_bac - len(all_strong_days_bac)}")
print("-"*50)
print("Trạm nghiên cứu khí tượng miền Nam")
# IN chi tiết từng ngày
print("=== CHI TIẾT BIẾN ĐỘNG THEO NGÀY ===\n")
for i in range(n_days_nam):
    print(f"Ngày {i} → {i+1}:")
    print(f"  ΔT = {delta_temp_nam[i]:+.2f}°C | ΔH = {delta_humid_nam[i]:+.2f}% | ΔR = {delta_rain_nam[i]:+.2f}mm")
    print(f"  Tình trạng: {tinh_trang_nam[i]}")
    print()

# IN tóm tắt
print("=== BIẾN ĐỘNG NHIỆT ĐỘ (> 8°C) ===")
if len(strong_temp_days_nam) > 0:
    for i in strong_temp_days_nam:
        print(f"Ngày {i} → {i+1}: ΔT = {delta_temp_nam[i]:.2f}°C")
else:
    print("Không có ngày nào biến động nhiệt độ > 8°C")

print("\n=== BIẾN ĐỘNG ĐỘ ẨM (> 15%) ===")
if len(strong_humid_days_nam) > 0:
    for i in strong_humid_days_nam:
        print(f"Ngày {i} → {i+1}: ΔH = {delta_humid_nam[i]:.2f}%")
else:
    print("Không có ngày nào biến động độ ẩm > 15%")

print("\n=== BIẾN ĐỘNG LƯỢNG MƯA (> 40mm) ===")
if len(strong_rain_days_nam) > 0:
    for i in strong_rain_days_nam:
        print(f"Ngày {i} → {i+1}: ΔR = {delta_rain_nam[i]:.2f}mm")
else:
    print("Không có ngày nào biến động lượng mưa > 40mm")

# Tìm những ngày có bất kỳ biến động mạnh nào
all_strong_days = np.unique(np.concatenate([strong_temp_days_nam, strong_humid_days_nam, strong_rain_days_nam]))
print(f"\n=== TỔNG HỢP ===")
print(f"Tổng ngày có biến động mạnh: {len(all_strong_days)}")
print(f"Tổng ngày bình thường: {n_days_bac - len(all_strong_days)}")
```
