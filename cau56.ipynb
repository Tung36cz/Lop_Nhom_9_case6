```python
import csv

header = ["Day", "Temperature", "Humidity", "Rainfall"]
rows = [
    [1, 30, 70, 2],
    [2, 31, 65, 0],
    [3, 29, 80, 5],
    [4, 28, 75, 10],
    [5, 32, 60, 0],
    [6, 33, 55, 1],
    [7, 31, 67, 0]
]

with open("weather.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(rows)

print("Đã tạo file weather.csv thành công!")
```

```python
#CÂU5
# Cột 0: Nhiệt độ T
# Cột 1: Độ ẩm H
# Cột 2: Lượng mưa R
import numpy as np    # Trung bình (mean) → xu hướng chung.

# Độ lệch chuẩn (std) → mức dao động mạnh hay yếu.

# Giá trị lớn nhất/nhỏ nhất (max/min).

# Xác định ngày xảy ra giá trị lớn nhất/nhỏ nhất (argmax/argmin)
#-->   thời tiết theo tháng/quý/năm.


data = np.genfromtxt("weather.csv", delimiter=",", skip_header=1)
temp = data[:, 1]     # nhiệt độ
humid = data[:, 2]    # độ ẩm
rain = data[:, 3]     # lượng mưa

#NHIỆT ĐỘ
mean_temp = temp.mean()
std_temp = temp.std()
min_temp = temp.min()
max_temp = temp.max()

day_min_temp = temp.argmin() + 1
day_max_temp = temp.argmax() + 1

#ĐỘ ẨM
mean_humid = humid.mean()
std_humid = humid.std()
min_humid = humid.min()
max_humid = humid.max()

day_min_humid = humid.argmin() + 1
day_max_humid = humid.argmax() + 1

#LƯỢNG MƯA
mean_rain = rain.mean()
std_rain = rain.std()
min_rain = rain.min()
max_rain = rain.max()

day_min_rain = rain.argmin() + 1
day_max_rain = rain.argmax() + 1

#IN KẾT QUẢ("=== THỐNG KÊ NHIỆT ĐỘ ===")
print("Nhiệt độ TB:", mean_temp)
print("Độ lệch chuẩn:", std_temp)
print("Thấp nhất:", min_temp, "ngày", day_min_temp)
print("Cao nhất:", max_temp, "ngày", day_max_temp)

print("\n=== THỐNG KÊ ĐỘ ẨM ===")
print("Độ ẩm TB:", mean_humid)
print("Độ lệch chuẩn:", std_humid)
print("Thấp nhất:", min_humid, "ngày", day_min_humid)
print("Cao nhất:", max_humid, "ngày", day_max_humid)

print("\n=== THỐNG KÊ LƯỢNG MƯA ===")
print("Mưa TB:", mean_rain)
print("Độ lệch chuẩn:", std_rain)
print("Thấp nhất:", min_rain, "ngày", day_min_rain)
print("Cao nhất:", max_rain, "ngày", day_max_rain)


#CÂU6

#DÙNG np.diff() TÌM NGÀY CÓ BIẾN ĐỘNG LỚN
import numpy as np

# np.diff() tính độ thay đổi giữa 2 ngày liên tiếp:

# Nếu nhiệt độ từ ngày i → i+1 tăng 5 độ → biến động mạnh.

# Nếu độ ẩm giảm 15% → cũng là biến động mạnh.

# Nếu lượng mưa tăng đột ngột → cảnh báo thời tiết. 
 
#Độ biến động từng yếu tố
delta_temp = np.diff(temp)      # biến động nhiệt độ
delta_humid = np.diff(humid)    # biến động độ ẩm
delta_rain = np.diff(rain)      # biến động lượng mưa


#Xác định ngưỡng biến động mạnh
# Nhiệt độ biến động mạnh nếu thay đổi ≥ 3°C

# Độ ẩm biến động mạnh nếu thay đổi ≥ 10%

# Mưa biến động mạnh nếu thay đổi ≥ 20 mm


T_thres = 3
H_thres = 10
R_thres = 20


#Tìm các ngày biến động mạnh

strong_temp_days = np.where(abs(delta_temp) >= T_thres)[0]
strong_humid_days = np.where(abs(delta_humid) >= H_thres)[0]
strong_rain_days = np.where(abs(delta_rain) >= R_thres)[0] 


#IN chi tiết từng biến động 

print("=== BIẾN ĐỘNG NHIỆT ĐỘ ===")
for i in strong_temp_days:
    print(f"Ngày {i} → {i+1}: ΔT = {delta_temp[i]:.2f}")

print("\n=== BIẾN ĐỘNG ĐỘ ẨM ===")
for i in strong_humid_days:
    print(f"Ngày {i} → {i+1}: ΔH = {delta_humid[i]:.2f}")

print("\n=== BIẾN ĐỘNG LƯỢNG MƯAp ===")
for i in strong_rain_days:
    print(f"Ngày {i} → {i+1}: ΔR = {delta_rain[i]:.2f}")
print(f"tung bien dong la: nhiet do luong mua do am ")
```
