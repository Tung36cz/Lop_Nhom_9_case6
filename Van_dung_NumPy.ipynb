```python
import numpy as np

# Dữ liệu khí tượng: mỗi hàng là 1 ngày
# Cột 0: temperature, cột 1: humidity, cột 2: rainfall
weather = np.array([
    [30, 70, 5],
    [32, 65, 0],
    [28, 80, 10],
    [27, 75, 0],
    [31, 60, 2],
    [29, 85, 20],
    [33, 55, 0],
])
temperature = weather[:, 0]
humidity = weather[:, 1]
rainfall = weather[:, 2]

print(temperature)
print(humidity)
print(rainfall)
#nhiệt độ trung bình 
np.mean(temperature)
#ngày có lượng mưa cao nhất 
np.max(rainfall)
#tổng lượng mưa cả tuần
np.sum(rainfall)
#nhiệt độ cao nhất 
np.max(temperature)
#lọc các ngày có mưa 
rain_days = weather[weather[:, 2] > 0]
print(rain_days)
#lọc ngày nóng 
hot_days = weather[weather[:, 0] > 30]
print(hot_days)
#thêm 1 cột chỉ số nhiệt bằng broadcasting
heat_index = temperature + humidity * 0.1
weather_with_index = np.column_stack((weather, heat_index))

print(weather_with_index)







```
